##### 1. Comprar apenas latas de 18 litros: (apenas latas inteiras)

##### 2. Comprar apenas galões de 3,6 litros: (apenas galoes inteiros)

```python
import math
def apenas_galoes(area): # Função que calculará a quantidade de galoes
    litros = area/6 # Quantos litros serão necessários para a área
    if litros%3.6 > 0: # Se o resto da divisão da quantidade de litros pela quantidade de litros de uma lata for maior que zero(Ex: litros/18 = 1,1)
        galoes = math.ceil(litros/3.6) # O número de inteiro de galoes não seria suficiente, logo será preciso arredondar a quantidade
    else:
        galoes = litros/3.6 # Se o resto da divisão for 0, o resultado é inteiro
        
    return galoes

print('Você precisará de {} galões.' .format(apenas_galoes(area)))
```
